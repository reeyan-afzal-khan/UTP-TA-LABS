## Part A — Tabular Q-learning in a 4×4 GridWorld

**Goal.** Define a small grid world and let an agent learn, from rewards alone, how to reach the goal.

*Questions 2–5 · Data: none: the notebook defines the grid world*

### Question 2 — Define the world

The world is a $4\times4$ grid. The agent starts top-left and must reach the goal bottom-right while avoiding a trap. Each move costs a little ($-0.04$), reaching the goal pays $+1$, and falling into the trap costs $-1$ and ends the episode. Walking into a wall leaves the agent where it is. That is the whole environment: a `step` function that takes a state and an action and returns the next state, the reward, and whether the episode is over.

```python
rng = np.random.default_rng(42)              # one random generator, so every run explores the same way

# The world: a 4 x 4 grid. Cells are (row, column), with (0, 0) at the top left.
ROWS, COLS = 4, 4
START, GOAL, TRAP = (0, 0), (3, 3), (1, 3)
ACTIONS = [(-1, 0), (1, 0), (0, -1), (0, 1)]        # how each move changes (row, column)
ACTION_NAMES = ["up", "down", "left", "right"]
ARROWS = {"up": (0, 1), "down": (0, -1), "left": (-1, 0), "right": (1, 0)}   # for drawing the policy

def step(state, action):
    """Apply one action. Returns (next_state, reward, done)."""
    r, c = state
    dr, dc = ACTIONS[action]
    # A move into the wall leaves the agent where it is.
    next_state = (min(max(r + dr, 0), ROWS - 1), min(max(c + dc, 0), COLS - 1))
    if next_state == GOAL:
        return next_state, 1.0, True
    if next_state == TRAP:
        return next_state, -1.0, True
    return next_state, -0.04, False          # every ordinary move costs a little, so shorter routes are better
```

### Question 3 — Train the Q-table

The Q-table has one number per (cell, action): the agent's current estimate of the long-run return from taking that action there. It starts at zero. Each step, the estimate is nudged toward what actually happened: the reward received plus the discounted value of the best move from the next cell. The nudge size is α; the discount γ says how much future reward is worth today. Early on the agent explores at random ($\varepsilon=1$); as ε decays it exploits what it has learned.

```python
Q = np.zeros((ROWS, COLS, len(ACTIONS)))     # Q[row, col, action]: the value of each move from each cell
alpha, gamma = 0.15, 0.95                    # learning rate, and how much a later reward is worth
epsilon, min_epsilon = 1.0, 0.05             # chance of a random move: starts at 1, shrinks to 0.05

def choose_action(state, explore=True):
    """Epsilon-greedy: a random move with probability epsilon, otherwise the best known move."""
    if explore and rng.random() < epsilon:
        return int(rng.integers(len(ACTIONS)))       # explore: random move
    return int(np.argmax(Q[state]))                  # exploit: best known move

training_returns = []
for episode in range(5000):
    state, episode_return = START, 0.0
    for _ in range(100):                             # at most 100 moves per episode
        action = choose_action(state)
        next_state, reward, done = step(state, action)
        episode_return += reward
        # The Q-learning update: move Q towards the reward plus the best value of the next cell.
        target = reward if done else reward + gamma * np.max(Q[next_state])
        Q[state][action] += alpha * (target - Q[state][action])
        state = next_state
        if done:
            break
    training_returns.append(episode_return)
    epsilon = max(min_epsilon, epsilon * 0.9985)     # explore a little less after every episode
```

### Question 4 — Read the training curve

The training curve is the return per episode, smoothed over 100 episodes. Early returns are negative because a random walker wanders into the trap or times out; as exploration decays the curve rises to the best achievable return.

```python
# The return of a single episode is noisy, so plot the mean over a sliding window of 100 episodes.
window = 100
moving = np.convolve(training_returns, np.ones(window) / window, mode="valid")
plt.figure(figsize=(6, 3.6))
plt.plot(np.arange(window - 1, len(training_returns)), moving)
plt.axhline(0.8, linestyle="--", color="grey", label="best possible return (6 moves)")
plt.xlabel("Episode")
plt.ylabel("Mean return over last 100 episodes")
plt.title("Q-learning training curve")
plt.legend()
plt.tight_layout()
plt.savefig(RESULTS / "q_learning_training_curve.png", dpi=160, bbox_inches="tight")
plt.show()
```

### Question 5 — Evaluate the greedy policy

Training returns mix learning with random moves, so they are not a fair measure of the policy. Evaluation switches exploration off and follows the greedy policy: in each cell, take the action with the highest Q-value.

```python
# Run the learned policy 200 times without exploring: always the best known move.
successes, returns = 0, []
for _ in range(200):
    state, total = START, 0.0
    for _ in range(100):
        state, reward, done = step(state, choose_action(state, explore=False))
        total += reward
        if done:
            successes += int(state == GOAL)
            break
    returns.append(total)

print("success rate:", successes / 200)
print("mean return:", round(float(np.mean(returns)), 3))
print("learned action at start:", ACTION_NAMES[choose_action(START, explore=False)])
```

**What to look for.** The shortest route is six moves: five at $-0.04$ and the sixth into the goal for $+1$, a return of $0.80$. A mean return of exactly $0.8$ with a 100% success rate means the agent found that route and never deviates from it.

**What the output shows.** The training curve climbs from negative returns (early wandering into the trap) to a plateau near 0.8. The greedy policy reaches the goal 200 times out of 200 with mean return 0.800, the best possible: five moves at −0.04 and a sixth into the goal for +1. There is no fixed dataset, no label, and no train/test split; the data depends on the policy being learned.

**Checkpoint: before you move on**

- The first 100 episodes average a return of $-0.914$, and the last 100 average 0.788.
- The greedy policy reaches the goal 200 times out of 200, with a return of 0.8.

## Part B — The Q-table is the model: save it and read the policy

**Goal.** Save the learned table, read the policy from it, and compare its values with the exact ones.

*Questions 6–7 · Data: the learned Q-table*

### Question 6 — Save the table and read the policy

Everything the agent learned is in the Q-table, so saving it is saving the model. Anyone with the file can act without retraining: in each cell, pick the action with the largest value.

```python
# The Q-table is the whole model: save it, load it back, and read the best move in every cell.
q_path = MODELS / "gridworld_q_table.npy"
np.save(q_path, Q)
Q_loaded = np.load(q_path)

def greedy_policy(Q_table, trap=TRAP):
    """The name of the best move in every cell, with the goal and the trap marked."""
    policy = np.full((ROWS, COLS), "", dtype=object)
    for r in range(ROWS):
        for c in range(COLS):
            if (r, c) == GOAL:
                policy[r, c] = "GOAL"
            elif (r, c) == trap:
                policy[r, c] = "TRAP"
            else:
                policy[r, c] = ACTION_NAMES[int(np.argmax(Q_table[r, c]))]
    return policy

policy = greedy_policy(Q_loaded)
print(policy)
```

### Question 7 — Draw the policy and the values

The policy is easier to read as arrows on the grid, drawn over a heat map of each cell's value (the best Q-value available there). Values rise towards the goal.

```python
def plot_policy(Q_table, title, trap=TRAP):
    """Color each cell by its best value, and draw an arrow for its best move."""
    values = Q_table.max(axis=2)
    plt.figure(figsize=(4.4, 4.2))
    plt.imshow(values, cmap="YlGn")
    plt.colorbar(label="value of the best action")
    for r in range(ROWS):
        for c in range(COLS):
            if (r, c) == GOAL:
                plt.text(c, r, "GOAL", ha="center", va="center", fontweight="bold")
            elif (r, c) == trap:
                plt.text(c, r, "TRAP", ha="center", va="center", color="red", fontweight="bold")
            else:
                dx, dy = ARROWS[ACTION_NAMES[int(np.argmax(Q_table[r, c]))]]
                plt.arrow(c, r, 0.3 * dx, -0.3 * dy, head_width=0.15, color="black")
    plt.xticks(range(COLS))
    plt.yticks(range(ROWS))
    plt.title(title)
    plt.tight_layout()
    file_name = re.sub(r"[^a-z0-9]+", "_", title.lower()).strip("_") + ".png"   # the title, as a file name
    plt.savefig(RESULTS / file_name, dpi=160, bbox_inches="tight")
    plt.show()

plot_policy(Q_loaded, "Greedy policy and state values")
```

**What to look for.** Every arrow points along a shortest path to the goal, and the arrows in the right-hand column steer away from the trap. The cell above the trap, (0, 3), is worth 0.666, less than its left neighbour's 0.743, because the way out from there starts by stepping back left.

**What the output shows.** The route runs down the left column to row 2, steps right, drops to the bottom row and follows it into the goal, steering clear of the trap at (1, 3); the value heat map rises towards the goal. A 640-byte file is enough to act without retraining.

**Checkpoint: before you move on**

- The route from the start takes six moves.
- The first row of values reads 0.593, 0.666, 0.743, 0.666, the exact values of the chapter.

Move the trap onto the bottom row, at (3, 1). **Prediction before retraining:** the learned route runs down the left column and along the bottom row, straight through the new trap, so the arrows in the left column should stop pointing down and the route should shift to the top row and the right-hand column (or through the middle), for the same six-move length.

### Question 10 — One solution

```python
def train_q(trap, step_cost=-0.04, gamma_=0.95, episodes=5000, seed=42):
    """Retrain from scratch on a grid with the given trap position and step cost; returns the new Q-table."""
    r = np.random.default_rng(seed)
    Q_new = np.zeros((ROWS, COLS, len(ACTIONS)))
    eps = 1.0

    def step_new(state, action):
        rr, cc = state
        dr, dc = ACTIONS[action]
        nxt = (min(max(rr + dr, 0), ROWS - 1), min(max(cc + dc, 0), COLS - 1))
        if nxt == GOAL:
            return nxt, 1.0, True
        if nxt == trap:
            return nxt, -1.0, True
        return nxt, step_cost, False

    for _ in range(episodes):
        state = START
        for _ in range(100):
            if r.random() < eps:
                action = int(r.integers(len(ACTIONS)))
            else:
                action = int(np.argmax(Q_new[state]))
            nxt, reward, done = step_new(state, action)
            target = reward if done else reward + gamma_ * np.max(Q_new[nxt])
            Q_new[state][action] += alpha * (target - Q_new[state][action])
            state = nxt
            if done:
                break
        eps = max(min_epsilon, eps * 0.9985)
    return Q_new

moved_trap = (3, 1)
Q_moved = train_q(trap=moved_trap)
print(f"Greedy policy with the trap at {moved_trap}:")
for row in greedy_policy(Q_moved, trap=moved_trap):
    print("".join(name.ljust(7) for name in row).rstrip())
plot_policy(Q_moved, "Trap moved to (3, 1)", trap=moved_trap)
```

**Check against the prediction.** The route now turns right at row 1, crosses the middle of the grid along row 2, and drops into the goal from the right-hand column; the cell directly beside the trap, (3, 0), points up, away from it. The right-hand column, which the original policy avoided, is now safe and its arrows point down. The path is still six moves, so the greedy return is still 0.8. The agent was never told the trap had moved; it relearned from the rewards alone, which is the whole point of learning by interaction rather than from a fixed dataset.
